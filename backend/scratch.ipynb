```python
from slides.xml_instructor import parse_xml_to_model
from pydantic import BaseModel, Field
from typing import List, Optional
```

```python
class SlideScript(BaseModel):
        slide_script: str = Field(..., min_length=1, max_length=10)
        
class Response(BaseModel):
    slide_scripts: List[SlideScript] = Field(..., min_items=5, max_items=5)
```

```python
test_xml = """
<slide_scripts>
<slide_script>slide 1</slide_script>
<slide_script>slide 2</slide_script>
<slide_script>slide 3</slide_script>
<slide_script>slide 4</slide_script>
<slide_script>slide 5</slide_script>
</slide_scripts>
"""
```

```python
model = parse_xml_to_model(test_xml, Response)
model.dict()
```

```python
scripts_strings = [script.slide_script for script in model.slide_scripts]
scripts_strings
```
